```python
# This cell defines all the settings for our entire project.
# We run this once at the start of every session.
# Think of it as the "control panel" for the whole system.

import os
from google.colab import userdata
from openai import OpenAI

# ── API SETUP ──────────────────────────────────────────────
# Load the Groq key from Colab secrets
os.environ["GROQ_API_KEY"] = userdata.get('GROQ_API_KEY')

# Create the AI client — every agent will use this same client
client = OpenAI(
    api_key=os.environ["GROQ_API_KEY"],
    base_url="https://api.groq.com/openai/v1"
)

# The AI model we are using — free and powerful
MODEL = "llama-3.3-70b-versatile"

# ── FOLDER PATHS ───────────────────────────────────────────
BASE        = "/content/drive/MyDrive/AI_Book_Writer"
INPUT_DIR   = f"{BASE}/input"       # Book title goes here
PROMPTS_DIR = f"{BASE}/prompts"     # Prompts sent to each agent
CHAPTERS_DIR= f"{BASE}/chapters"    # Raw chapter drafts
EDITED_DIR  = f"{BASE}/edited"      # Edited chapters
OUTPUT_DIR  = f"{BASE}/output"      # Final DOCX
DATABASE_DIR= f"{BASE}/database"    # SQLite database
LOGS_DIR    = f"{BASE}/logs"        # Progress logs
AGENTS_DIR  = f"{BASE}/agents"      # Agent code

print("✅ Config loaded")
print(f"   Model  : {MODEL}")
print(f"   Base   : {BASE}")
```

```python
# This is a helper function that every agent will use.
# Instead of writing the same API call 8 times (once per agent),
# we write it once here and just call it by name later.

def ask_ai(system_prompt, user_prompt, max_tokens=2000):
    """
    Sends a message to the AI and returns the response as a string.

    system_prompt : tells the AI what ROLE to play (e.g. "You are a researcher")
    user_prompt   : the actual task or question
    max_tokens    : maximum length of the response
    """

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": system_prompt},  # AI's role/personality
            {"role": "user",   "content": user_prompt}     # The actual task
        ],
        max_tokens=max_tokens,
        temperature=0.7   # 0 = very focused, 1 = more creative. 0.7 is a good balance
    )

    # Extract just the text from the response object
    return response.choices[0].message.content


# ── Quick test ─────────────────────────────────────────────
test = ask_ai(
    system_prompt="You are a helpful assistant.",
    user_prompt="Name three famous scientists in one sentence."
)
print("Test response:", test)
print("\n✅ Helper function working!")
```

```python
def outline_agent(book_title, research_notes):
    """
    Takes the research notes and creates a full chapter-by-chapter outline.
    This is Agent #2 in our pipeline.
    """

    print(f"\n📋 Outline Agent starting...")
    print(f"   Book: {book_title}")

    system_prompt = """You are an expert book editor and outline specialist.
You create clear, detailed, well-structured book outlines.
Every chapter must have a clear purpose and flow logically into the next.
You write outlines for non-fiction books that are practical and engaging."""

    user_prompt = f"""Using the research notes below, create a detailed outline
for a 10-chapter non-fiction book.

BOOK TITLE: {book_title}

RESEARCH NOTES:
{research_notes}

For each of the 10 chapters provide:
- CHAPTER NUMBER and TITLE
- PURPOSE (one sentence — what this chapter achieves)
- KEY POINTS (4-5 bullet points of what will be covered)
- OPENING HOOK (one sentence to start the chapter)
- CLOSING TAKEAWAY (what the reader will walk away knowing)

Make sure the chapters flow logically from beginning to end.
The book should take the reader on a complete journey."""

    outline = ask_ai(system_prompt, user_prompt, max_tokens=4000)

    # Save to Google Drive
    safe_title = book_title.replace(" ", "_").lower()
    output_path = f"{BASE}/prompts/outline_{safe_title}.txt"

    with open(output_path, "w", encoding="utf-8") as f:
        f.write(f"BOOK TITLE: {book_title}\n")
        f.write("=" * 60 + "\n\n")
        f.write(outline)

    print(f"✅ Outline complete!")
    print(f"   Saved to: {output_path}")
    print(f"   Length: {len(outline)} characters")

    return outline


# ── RUN IT ─────────────────────────────────────────────────
# We pass in the research we already generated
outline = outline_agent(BOOK_TITLE, research)

# Preview
print("\n── PREVIEW (first 800 chars) ──────────────────────────")
print(outline[:800])
```

```python
def chapter_writer_agent(book_title, outline, chapter_number):
    """
    Writes one full chapter based on the outline.
    We call this once per chapter (10 times total).

    chapter_number : integer 1 through 10
    """

    print(f"\n✍️  Chapter Writer Agent starting...")
    print(f"   Book   : {book_title}")
    print(f"   Chapter: {chapter_number} of 10")

    system_prompt = """You are a professional non-fiction book author with 20 years
of experience writing bestselling books.
Your writing is clear, engaging, practical, and human.
You use real examples, stories, and analogies to explain ideas.
You never use filler words or repetitive sentences.
Each chapter feels complete and leaves the reader wanting more."""

    user_prompt = f"""Write Chapter {chapter_number} of the book titled "{book_title}".

Use this outline as your guide:
{outline}

INSTRUCTIONS:
- Write ONLY Chapter {chapter_number} — not other chapters
- Length: 800 to 1200 words
- Start with an engaging opening paragraph
- Use subheadings to organize the content
- Include at least one real-world example or story
- End with a strong summary and one actionable tip
- Write in second person ("you") to speak directly to the reader
- Tone: warm, practical, motivating

Begin writing the chapter now. Do not include any commentary
or notes — just the chapter content itself."""

    chapter_text = ask_ai(system_prompt, user_prompt, max_tokens=4000)

    # Save to Google Drive
    safe_title = book_title.replace(" ", "_").lower()
    filename = f"chapter_{chapter_number:02d}_{safe_title}.txt"
    # chapter_number:02d means: format as 2 digits, so chapter 1 = "01"
    # This keeps files in order: chapter_01, chapter_02 ... chapter_10
    output_path = f"{CHAPTERS_DIR}/{filename}"

    with open(output_path, "w", encoding="utf-8") as f:
        f.write(f"BOOK: {book_title}\n")
        f.write(f"CHAPTER: {chapter_number}\n")
        f.write("=" * 60 + "\n\n")
        f.write(chapter_text)

    print(f"✅ Chapter {chapter_number} complete!")
    print(f"   Saved to: {output_path}")
    print(f"   Length: {len(chapter_text)} characters")

    return chapter_text


# ── RUN IT — Write Chapter 1 first ─────────────────────────
chapter_1 = chapter_writer_agent(BOOK_TITLE, outline, chapter_number=1)

# Preview first 600 characters
print("\n── PREVIEW (first 600 chars) ──────────────────────────")
print(chapter_1[:600])
```

```python
# This loop calls the chapter_writer_agent 10 times
# Once for each chapter in the book
# We store all chapters in a list so we can use them later

print(f"📚 Writing all 10 chapters of '{BOOK_TITLE}'")
print("   This will take 3-5 minutes. Please wait...\n")

# This list will hold all 10 chapter texts
all_chapters = []

# Loop from 1 to 10 inclusive
for chapter_num in range(1, 11):

    # Write this chapter
    chapter_text = chapter_writer_agent(BOOK_TITLE, outline, chapter_num)

    # Add it to our list
    all_chapters.append(chapter_text)

    # Small pause between chapters so we don't hit API rate limits
    # Rate limit = maximum requests per minute allowed by Groq free tier
    import time
    time.sleep(3)   # Wait 3 seconds before writing the next chapter

print("\n" + "=" * 50)
print(f"✅ All 10 chapters written!")
print(f"   Total chapters : {len(all_chapters)}")
print(f"   Total characters: {sum(len(c) for c in all_chapters)}")
print(f"   Check your Drive: {CHAPTERS_DIR}")
```

```python
def editor_agent(book_title, chapter_text, chapter_number):
    """
    Takes a raw chapter and improves it.
    Fixes flow, clarity, transitions, and engagement.
    This is Agent #4 in our pipeline.
    """

    print(f"\n📝 Editor Agent starting...")
    print(f"   Editing Chapter {chapter_number}...")

    system_prompt = """You are a senior book editor with 25 years of experience
editing bestselling non-fiction books.
You improve writing without changing the author's voice or meaning.
You focus on clarity, flow, engagement, and readability.
You never add new facts — you only improve what is already there."""

    user_prompt = f"""Edit and improve the following chapter from the book "{book_title}".

CHAPTER {chapter_number} — ORIGINAL DRAFT:
{chapter_text}

YOUR EDITING TASKS:
1. CLARITY — Make every sentence clear and easy to understand
2. FLOW — Ensure paragraphs connect smoothly with good transitions
3. ENGAGEMENT — Make the opening and closing stronger
4. CONCISENESS — Remove any repetitive or filler sentences
5. SUBHEADINGS — Make sure subheadings are clear and compelling
6. EXAMPLES — Strengthen any weak examples or stories
7. ACTIONABLE — Make the practical advice more specific and useful

Return the complete edited chapter.
Do not add commentary or editing notes — just the improved chapter."""

    edited_text = ask_ai(system_prompt, user_prompt, max_tokens=4000)

    # Save to edited folder
    safe_title = book_title.replace(" ", "_").lower()
    filename = f"edited_{chapter_number:02d}_{safe_title}.txt"
    output_path = f"{EDITED_DIR}/{filename}"

    with open(output_path, "w", encoding="utf-8") as f:
        f.write(f"BOOK: {book_title}\n")
        f.write(f"CHAPTER: {chapter_number} — EDITED\n")
        f.write("=" * 60 + "\n\n")
        f.write(edited_text)

    print(f"✅ Chapter {chapter_number} edited!")
    print(f"   Saved to: {output_path}")
    print(f"   Length: {len(edited_text)} characters")

    return edited_text


# ── Edit all 10 chapters ────────────────────────────────────
import time

print(f"📝 Editing all 10 chapters of '{BOOK_TITLE}'")
print("   This will take 3-5 minutes...\n")

all_edited_chapters = []

for i, chapter_text in enumerate(all_chapters, start=1):
    # enumerate gives us both the index (i) and the value (chapter_text)

    edited = editor_agent(BOOK_TITLE, chapter_text, i)
    all_edited_chapters.append(edited)
    time.sleep(3)

print("\n" + "=" * 50)
print(f"✅ All 10 chapters edited!")
print(f"   Total characters: {sum(len(c) for c in all_edited_chapters)}")
print(f"   Check your Drive: {EDITED_DIR}")
```

```python
import docx
from docx import Document
from docx.shared import Pt, Inches
from docx.enum.text import WD_ALIGN_PARAGRAPH

def assembly_agent(book_title, edited_chapters):
    """
    Takes all edited chapters and assembles them into
    a properly formatted Microsoft Word .docx file.
    This is the final agent in our pipeline.
    """

    print(f"\n📦 Assembly Agent starting...")
    print(f"   Assembling '{book_title}'...")

    # Create a brand new Word document
    doc = Document()

    # ── TITLE PAGE ─────────────────────────────────────────
    # Add the book title in large centered text
    title_para = doc.add_paragraph()
    title_para.alignment = WD_ALIGN_PARAGRAPH.CENTER  # Center align
    title_run = title_para.add_run(book_title)
    title_run.bold = True
    title_run.font.size = Pt(28)   # 28 point font — large title

    # Add author line below title
    author_para = doc.add_paragraph()
    author_para.alignment = WD_ALIGN_PARAGRAPH.CENTER
    author_run = author_para.add_run("\nBy AI Book Writer System")
    author_run.font.size = Pt(14)

    # Add a page break after the title page
    # This forces the next content onto a new page
    doc.add_page_break()

    # ── TABLE OF CONTENTS PAGE ─────────────────────────────
    toc_heading = doc.add_paragraph()
    toc_run = toc_heading.add_run("Table of Contents")
    toc_run.bold = True
    toc_run.font.size = Pt(18)

    # Add a line for each chapter in the TOC
    for i in range(1, 11):
        toc_line = doc.add_paragraph()
        toc_line.add_run(f"  Chapter {i}").bold = True

    doc.add_page_break()

    # ── CHAPTERS ───────────────────────────────────────────
    for i, chapter_text in enumerate(edited_chapters, start=1):

        print(f"   Adding Chapter {i}...")

        # Split the chapter into individual lines
        lines = chapter_text.strip().split('\n')

        for line in lines:
            line = line.strip()  # Remove extra spaces at start/end

            # Skip empty lines
            if not line:
                doc.add_paragraph()
                continue

            # Detect markdown headings (## or ###) and format as headings
            if line.startswith('## '):
                # Remove the ## and add as a chapter heading
                heading_text = line.replace('## ', '').replace('# ', '')
                heading = doc.add_paragraph()
                heading.alignment = WD_ALIGN_PARAGRAPH.LEFT
                run = heading.add_run(heading_text)
                run.bold = True
                run.font.size = Pt(18)   # Chapter title size

            elif line.startswith('### '):
                # Subheading — smaller than chapter title
                heading_text = line.replace('### ', '')
                heading = doc.add_paragraph()
                run = heading.add_run(heading_text)
                run.bold = True
                run.font.size = Pt(14)   # Subheading size

            elif line.startswith('**') and line.endswith('**'):
                # Bold text — remove the ** markers
                bold_text = line.replace('**', '')
                para = doc.add_paragraph()
                run = para.add_run(bold_text)
                run.bold = True

            elif line.startswith('- ') or line.startswith('* '):
                # Bullet points
                bullet_text = line[2:]   # Remove the "- " or "* " prefix
                doc.add_paragraph(bullet_text, style='List Bullet')

            else:
                # Regular paragraph text
                para = doc.add_paragraph(line)
                para.paragraph_format.space_after = Pt(6)  # Small gap after each paragraph

        # Add a page break after each chapter
        doc.add_page_break()

    # ── SAVE THE DOCUMENT ──────────────────────────────────
    safe_title = book_title.replace(" ", "_").lower()
    output_path = f"{OUTPUT_DIR}/{safe_title}_final.docx"

    doc.save(output_path)

    print(f"\n✅ Book assembled successfully!")
    print(f"   Saved to: {output_path}")
    print(f"   Chapters : {len(edited_chapters)}")

    return output_path


# ── RUN IT ─────────────────────────────────────────────────
final_path = assembly_agent(BOOK_TITLE, all_edited_chapters)

print(f"\n🎉 YOUR BOOK IS COMPLETE!")
print(f"   Title : {BOOK_TITLE}")
print(f"   File  : {final_path}")
print(f"   Open your Google Drive to download it!")
```

```python
from docx import Document
from docx.shared import Pt, Twips, RGBColor
from docx.enum.text import WD_ALIGN_PARAGRAPH
from docx.oxml.ns import qn
from docx.oxml import OxmlElement

def assembly_agent(book_title, edited_chapters):

    print(f"\n📦 Assembly Agent starting...")
    print(f"   Building KDP-ready document...")

    doc = Document()

    # ── PAGE SIZE: KDP 6x9 inches ──────────────────────────
    # 1 inch = 1440 twips
    section = doc.sections[0]
    section.page_width        = Twips(8640)   # 6 inches
    section.page_height       = Twips(12960)  # 9 inches
    section.top_margin        = Twips(1008)   # 0.70 inch
    section.bottom_margin     = Twips(1008)   # 0.70 inch
    section.left_margin       = Twips(1080)   # 0.75 inch (gutter for binding)
    section.right_margin      = Twips(864)    # 0.60 inch
    section.header_distance   = Twips(720)
    section.footer_distance   = Twips(720)

    # ── DEFAULT FONT ───────────────────────────────────────
    # Garamond is a classic book font accepted by KDP
    style = doc.styles['Normal']
    style.font.name = 'Garamond'
    style.font.size = Pt(12)

    # ── HELPER: body paragraph (justified + first-line indent) ──
    def add_body(text, first_para=False):
        p = doc.add_paragraph()
        p.alignment = WD_ALIGN_PARAGRAPH.JUSTIFY  # Full justification
        pPr = p._p.get_or_add_pPr()
        # Correct XML order: spacing first, then ind, then jc
        sp = OxmlElement('w:spacing')
        sp.set(qn('w:line'), '276')       # 1.15 line spacing
        sp.set(qn('w:lineRule'), 'auto')
        sp.set(qn('w:before'), '0')
        sp.set(qn('w:after'), '0')        # No gap between paragraphs
        pPr.append(sp)
        if not first_para:
            ind = OxmlElement('w:ind')
            ind.set(qn('w:firstLine'), '720')  # 0.5 inch first-line indent
            pPr.append(ind)
        jc = OxmlElement('w:jc')
        jc.set(qn('w:val'), 'both')       # Full justify
        pPr.append(jc)
        run = p.add_run(text)
        run.font.name = 'Garamond'
        run.font.size = Pt(12)
        return p

    # ── HELPER: subheading ─────────────────────────────────
    def add_subheading(text):
        p = doc.add_paragraph()
        p.alignment = WD_ALIGN_PARAGRAPH.LEFT
        pPr = p._p.get_or_add_pPr()
        sp = OxmlElement('w:spacing')
        sp.set(qn('w:before'), '360')
        sp.set(qn('w:after'), '120')
        pPr.append(sp)
        run = p.add_run(text)
        run.font.name = 'Garamond'
        run.font.size = Pt(13)
        run.bold = True

    # ── HELPER: chapter title ──────────────────────────────
    def add_chapter_title(number, title):
        doc.add_page_break()
        # "Chapter X" label
        lbl = doc.add_paragraph()
        lbl.alignment = WD_ALIGN_PARAGRAPH.CENTER
        pPr = lbl._p.get_or_add_pPr()
        sp = OxmlElement('w:spacing')
        sp.set(qn('w:before'), '1440')
        sp.set(qn('w:after'), '240')
        pPr.append(sp)
        r = lbl.add_run(f"Chapter {number}")
        r.font.name = 'Garamond'
        r.font.size = Pt(12)
        r.font.color.rgb = RGBColor(0x66, 0x66, 0x66)
        # Title
        ttl = doc.add_paragraph()
        ttl.alignment = WD_ALIGN_PARAGRAPH.CENTER
        pPr2 = ttl._p.get_or_add_pPr()
        sp2 = OxmlElement('w:spacing')
        sp2.set(qn('w:before'), '0')
        sp2.set(qn('w:after'), '720')
        pPr2.append(sp2)
        r2 = ttl.add_run(title)
        r2.font.name = 'Garamond'
        r2.font.size = Pt(20)
        r2.bold = True

    # ── PAGE NUMBERS IN FOOTER ─────────────────────────────
    def add_page_numbers():
        footer = doc.sections[0].footer
        fp = footer.paragraphs[0]
        fp.alignment = WD_ALIGN_PARAGRAPH.CENTER
        fp.clear()
        run = fp.add_run()
        run.font.name = 'Garamond'
        run.font.size = Pt(10)
        for tag, val in [('w:fldChar','begin'), ('w:instrText','PAGE'), ('w:fldChar','end')]:
            el = OxmlElement(tag)
            if tag == 'w:instrText':
                el.text = val
            else:
                el.set(qn('w:fldCharType'), val)
            run._r.append(el)

    # ── TITLE PAGE ─────────────────────────────────────────
    tp = doc.add_paragraph()
    tp.alignment = WD_ALIGN_PARAGRAPH.CENTER
    pPr = tp._p.get_or_add_pPr()
    sp = OxmlElement('w:spacing')
    sp.set(qn('w:before'), '2880')
    sp.set(qn('w:after'), '360')
    pPr.append(sp)
    tr = tp.add_run(book_title)
    tr.font.name = 'Garamond'
    tr.font.size = Pt(28)
    tr.bold = True

    ap = doc.add_paragraph()
    ap.alignment = WD_ALIGN_PARAGRAPH.CENTER
    pPr2 = ap._p.get_or_add_pPr()
    sp2 = OxmlElement('w:spacing')
    sp2.set(qn('w:before'), '2880')
    pPr2.append(sp2)
    ar = ap.add_run("By AI Book Writer")
    ar.font.name = 'Garamond'
    ar.font.size = Pt(13)

    # ── COPYRIGHT PAGE ─────────────────────────────────────
    doc.add_page_break()
    cp = doc.add_paragraph()
    pPr3 = cp._p.get_or_add_pPr()
    sp3 = OxmlElement('w:spacing')
    sp3.set(qn('w:before'), '8640')
    pPr3.append(sp3)
    cr = cp.add_run(
        f"Copyright © 2025 {book_title}. All rights reserved.\n"
        "No part of this publication may be reproduced without permission."
    )
    cr.font.name = 'Garamond'
    cr.font.size = Pt(9)
    cr.font.color.rgb = RGBColor(0x55, 0x55, 0x55)

    # ── TABLE OF CONTENTS ──────────────────────────────────
    doc.add_page_break()
    toc_h = doc.add_paragraph()
    toc_h.alignment = WD_ALIGN_PARAGRAPH.CENTER
    pPr4 = toc_h._p.get_or_add_pPr()
    sp4 = OxmlElement('w:spacing')
    sp4.set(qn('w:before'), '720')
    sp4.set(qn('w:after'), '480')
    pPr4.append(sp4)
    th = toc_h.add_run("Contents")
    th.font.name = 'Garamond'
    th.font.size = Pt(16)
    th.bold = True

    # ── CHAPTERS ───────────────────────────────────────────
    chapter_titles = []  # collect titles for TOC

    for i, chapter_text in enumerate(edited_chapters, start=1):
        print(f"   Adding Chapter {i}...")
        lines = chapter_text.strip().split('\n')
        first_title = None
        first_body = True

        for line in lines:
            line = line.strip()
            if not line:
                continue

            # Grab first heading as chapter title
            if line.startswith('## ') or line.startswith('# '):
                clean = line.lstrip('#').strip()
                if first_title is None:
                    first_title = clean
                    chapter_titles.append(clean)
                    add_chapter_title(i, clean)
                else:
                    add_subheading(clean)
                first_body = True

            elif line.startswith('### '):
                add_subheading(line.replace('### ', ''))
                first_body = True

            elif line.startswith('**') and line.endswith('**'):
                add_subheading(line.replace('**', ''))
                first_body = True

            elif line.startswith('- ') or line.startswith('* '):
                # Bullet point — treat as body with dash
                add_body('• ' + line[2:], first_para=True)

            else:
                add_body(line, first_para=first_body)
                first_body = False

        # Fallback if no ## heading found
        if first_title is None:
            chapter_titles.append(f"Chapter {i}")

    # ── FILL TOC NOW THAT WE HAVE TITLES ──────────────────
    for i, title in enumerate(chapter_titles, 1):
        tp2 = doc.add_paragraph()
        # Insert before footer — we'll add TOC entries right after heading
        # (TOC was added to doc already; these go at end — acceptable for now)
        tp2.alignment = WD_ALIGN_PARAGRAPH.LEFT
        pPr5 = tp2._p.get_or_add_pPr()
        sp5 = OxmlElement('w:spacing')
        sp5.set(qn('w:before'), '60')
        sp5.set(qn('w:after'), '60')
        pPr5.append(sp5)
        nr = tp2.add_run(f"Chapter {i}   ")
        nr.font.name = 'Garamond'
        nr.font.size = Pt(11)
        nr.bold = True
        tr3 = tp2.add_run(title)
        tr3.font.name = 'Garamond'
        tr3.font.size = Pt(11)

    # Move TOC entries to right position (after TOC heading)
    # Find TOC heading paragraph index and insert after it
    toc_idx = None
    for idx, p in enumerate(doc.paragraphs):
        if p.text.strip() == 'Contents':
            toc_idx = idx
            break

    if toc_idx is not None:
        body = doc._body
        toc_para = doc.paragraphs[toc_idx]._p
        # Move the last N paragraphs (our TOC entries) to after toc_para
        n = len(chapter_titles)
        all_p = body.findall(qn('w:p'))
        toc_entries = all_p[-n:]
        for entry in toc_entries:
            body.remove(entry)
            toc_para.addnext(entry)
            toc_para = entry  # next entry goes after this one

    # ── PAGE NUMBERS ───────────────────────────────────────
    add_page_numbers()

    # ── SAVE ───────────────────────────────────────────────
    safe_title = book_title.replace(" ", "_").lower()
    output_path = f"{OUTPUT_DIR}/{safe_title}_KDP.docx"
    doc.save(output_path)

    total_words = sum(len(p.text.split()) for p in doc.paragraphs)

    print(f"\n✅ KDP Book assembled!")
    print(f"   Saved to    : {output_path}")
    print(f"   Total words : {total_words}")
    print(f"   Est. pages  : ~{total_words // 250} at 6x9 KDP size")
    print(f"   Format      : 6x9 inch, Garamond 12pt, justified, page numbers")

    return output_path


# ── RUN IT ─────────────────────────────────────────────────
final_path = assembly_agent(BOOK_TITLE, all_edited_chapters)

print(f"\n🎉 YOUR KDP-READY BOOK IS COMPLETE!")
print(f"   Download from: Google Drive → AI_Book_Writer → output")
```

```python
# ════════════════════════════════════════════════
# ✏️  CHANGE ONLY THIS LINE — then click Run ▶
BOOK_TITLE = "Live to 100: The Science of Reversing Your Age"
# ════════════════════════════════════════════════

# ── Install ──────────────────────────────────────
import subprocess, sys
subprocess.run([sys.executable,"-m","pip","install","-q",
                "python-docx","openai"], check=True)

# ── Mount Drive ───────────────────────────────────
from google.colab import drive, userdata
drive.mount('/content/drive')

# ── Folders ───────────────────────────────────────
import os, time
BASE         = "/content/drive/MyDrive/AI_Book_Writer"
PROMPTS_DIR  = f"{BASE}/prompts"
CHAPTERS_DIR = f"{BASE}/chapters"
EDITED_DIR   = f"{BASE}/edited"
OUTPUT_DIR   = f"{BASE}/output"
for p in [PROMPTS_DIR,CHAPTERS_DIR,EDITED_DIR,OUTPUT_DIR]:
    os.makedirs(p, exist_ok=True)
print("✅ Folders ready")

# ── API Client ────────────────────────────────────
from openai import OpenAI
client = OpenAI(
    api_key=userdata.get('GROQ_API_KEY'),
    base_url="https://api.groq.com/openai/v1"
)
MODEL = "llama-3.3-70b-versatile"

def ask_ai(system_prompt, user_prompt, max_tokens=2000):
    r = client.chat.completions.create(
        model=MODEL,
        messages=[{"role":"system","content":system_prompt},
                  {"role":"user","content":user_prompt}],
        max_tokens=max_tokens, temperature=0.7)
    return r.choices[0].message.content

safe = BOOK_TITLE.replace(" ","_").lower()
print(f"📖 Book: {BOOK_TITLE}\n")

# ── 1. Research ───────────────────────────────────
print("🔍 Research Agent...")
research = ask_ai(
    "You are an expert book researcher. Be factual and thorough.",
    f"""Research this book topic: {BOOK_TITLE}
Cover: 1) Overview & why it matters  2) 8-10 key concepts
3) Target audience  4) 10 chapter themes  5) Key insights
6) Best tone & style. Be thorough — this drives a full book.""",
    max_tokens=3000)
with open(f"{PROMPTS_DIR}/research_{safe}.txt","w") as f:
    f.write(research)
print(f"✅ Research done ({len(research):,} chars)\n")

# ── 2. Outline ────────────────────────────────────
print("📋 Outline Agent...")
outline = ask_ai(
    "You are an expert book editor. Create clear, detailed outlines.",
    f"""Create a 10-chapter outline for: {BOOK_TITLE}
Research notes:
{research}
For each chapter give: number & title, purpose, 4-5 key points,
opening hook, closing takeaway. Chapters must flow logically.""",
    max_tokens=4000)
with open(f"{PROMPTS_DIR}/outline_{safe}.txt","w") as f:
    f.write(outline)
print(f"✅ Outline done ({len(outline):,} chars)\n")

# ── 3. Write chapters ─────────────────────────────
print("✍️  Chapter Writer Agent (10 chapters — ~10 min)...")
all_chapters = []
for n in range(1, 11):
    print(f"   Writing chapter {n}/10...", end=" ")
    ch = ask_ai(
        "You are a professional non-fiction author. Clear, engaging, practical.",
        f"""Write Chapter {n} of "{BOOK_TITLE}".
Outline reference:
{outline}
RULES: Write ONLY chapter {n} · 4,000–5,000 words
· Start with engaging opening · Use ## subheadings
· Real-world examples & stories · End with summary + 1 action tip
· Second person (you) · Warm, practical, motivating tone
Write the full chapter now. No commentary.""",
        max_tokens=8000)
    all_chapters.append(ch)
    with open(f"{CHAPTERS_DIR}/chapter_{n:02d}_{safe}.txt","w") as f:
        f.write(ch)
    print(f"{len(ch):,} chars")
    time.sleep(5)
print(f"✅ All chapters written — {sum(len(c) for c in all_chapters):,} total chars\n")

# ── 4. Edit chapters ──────────────────────────────
print("📝 Editor Agent...")
all_edited = []
for i, ch in enumerate(all_chapters, 1):
    print(f"   Editing chapter {i}/10...", end=" ")
    ed = ask_ai(
        "You are a senior book editor. Improve without changing the author's voice.",
        f"""Edit Chapter {i} of "{BOOK_TITLE}":
{ch}
Tasks: 1) Clarity  2) Flow  3) Stronger opening/closing
4) Remove filler  5) Strengthen examples
Return the complete edited chapter. No commentary.""",
        max_tokens=8000)
    all_edited.append(ed)
    with open(f"{EDITED_DIR}/edited_{i:02d}_{safe}.txt","w") as f:
        f.write(ed)
    print(f"{len(ed):,} chars")
    time.sleep(5)
print(f"✅ All chapters edited — {sum(len(c) for c in all_edited):,} total chars\n")

# ── 5. Assemble KDP docx ─────────────────────────
print("📦 Assembly Agent — building KDP docx...")
from docx import Document
from docx.shared import Pt, Twips, RGBColor
from docx.enum.text import WD_ALIGN_PARAGRAPH
from docx.oxml.ns import qn
from docx.oxml import OxmlElement

doc = Document()
sec = doc.sections[0]
sec.page_width=Twips(8640); sec.page_height=Twips(12960)
sec.top_margin=Twips(1008); sec.bottom_margin=Twips(1008)
sec.left_margin=Twips(1080); sec.right_margin=Twips(864)
doc.styles['Normal'].font.name='Garamond'
doc.styles['Normal'].font.size=Pt(12)

def sp_el(before=0,after=0,line=None):
    s=OxmlElement('w:spacing')
    if before: s.set(qn('w:before'),str(before))
    if after:  s.set(qn('w:after'),str(after))
    if line:   s.set(qn('w:line'),str(line)); s.set(qn('w:lineRule'),'auto')
    return s

def add_body(text, first=False):
    p=doc.add_paragraph(); p.alignment=WD_ALIGN_PARAGRAPH.JUSTIFY
    pr=p._p.get_or_add_pPr(); pr.append(sp_el(0,0,276))
    if not first:
        ind=OxmlElement('w:ind'); ind.set(qn('w:firstLine'),'720'); pr.append(ind)
    r=p.add_run(text); r.font.name='Garamond'; r.font.size=Pt(12)

def add_sub(text):
    p=doc.add_paragraph(); p.alignment=WD_ALIGN_PARAGRAPH.LEFT
    p._p.get_or_add_pPr().append(sp_el(360,120))
    r=p.add_run(text); r.font.name='Garamond'; r.font.size=Pt(13); r.bold=True

def add_chapter_title(num, title):
    doc.add_page_break()
    lbl=doc.add_paragraph(); lbl.alignment=WD_ALIGN_PARAGRAPH.CENTER
    lbl._p.get_or_add_pPr().append(sp_el(1440,240))
    r=lbl.add_run(f"Chapter {num}")
    r.font.name='Garamond'; r.font.size=Pt(12)
    r.font.color.rgb=RGBColor(0x66,0x66,0x66)
    ttl=doc.add_paragraph(); ttl.alignment=WD_ALIGN_PARAGRAPH.CENTER
    ttl._p.get_or_add_pPr().append(sp_el(0,720))
    r2=ttl.add_run(title); r2.font.name='Garamond'
    r2.font.size=Pt(20); r2.bold=True

# Title page
tp=doc.add_paragraph(); tp.alignment=WD_ALIGN_PARAGRAPH.CENTER
tp._p.get_or_add_pPr().append(sp_el(2880,360))
tr=tp.add_run(BOOK_TITLE); tr.font.name='Garamond'
tr.font.size=Pt(28); tr.bold=True
ap=doc.add_paragraph(); ap.alignment=WD_ALIGN_PARAGRAPH.CENTER
ap._p.get_or_add_pPr().append(sp_el(2880,0))
ap.add_run("By AI Book Writer").font.size=Pt(13)

# Copyright
doc.add_page_break()
cp=doc.add_paragraph()
cp._p.get_or_add_pPr().append(sp_el(8640,0))
cr=cp.add_run("Copyright © 2025. All rights reserved.\n"
    "No part of this publication may be reproduced without permission.")
cr.font.size=Pt(9); cr.font.color.rgb=RGBColor(0x55,0x55,0x55)

# TOC placeholder
doc.add_page_break()
toc_h=doc.add_paragraph(); toc_h.alignment=WD_ALIGN_PARAGRAPH.CENTER
toc_h._p.get_or_add_pPr().append(sp_el(720,480))
th=toc_h.add_run("Contents"); th.font.name='Garamond'
th.font.size=Pt(16); th.bold=True
toc_anchor=toc_h._p

# Chapters
chapter_titles=[]
for i, ch in enumerate(all_edited, 1):
    print(f"   Adding chapter {i}...")
    lines=ch.strip().split('\n'); first_title=None; first_body=True
    for line in lines:
        line=line.strip()
        if not line: continue
        if line.startswith('## ') or line.startswith('# '):
            clean=line.lstrip('#').strip()
            if first_title is None:
                first_title=clean; chapter_titles.append(clean)
                add_chapter_title(i,clean)
            else: add_sub(clean)
            first_body=True
        elif line.startswith('### '): add_sub(line[4:]); first_body=True
        elif line.startswith('**') and line.endswith('**'):
            add_sub(line.replace('**','')); first_body=True
        elif line.startswith('- ') or line.startswith('* '):
            add_body('• '+line[2:], first=True)
        else:
            add_body(line, first=first_body); first_body=False
    if first_title is None: chapter_titles.append(f"Chapter {i}")

# Insert TOC entries
ins=toc_anchor
for i, title in enumerate(chapter_titles, 1):
    te=doc.add_paragraph(); te.alignment=WD_ALIGN_PARAGRAPH.LEFT
    te._p.get_or_add_pPr().append(sp_el(60,60))
    r1=te.add_run(f"Chapter {i}   "); r1.font.name='Garamond'
    r1.font.size=Pt(11); r1.bold=True
    r2=te.add_run(title); r2.font.name='Garamond'; r2.font.size=Pt(11)
    el=te._p; el.getparent().remove(el); ins.addnext(el); ins=el

# Page numbers
fp=doc.sections[0].footer.paragraphs[0]
fp.alignment=WD_ALIGN_PARAGRAPH.CENTER; fp.clear()
run=fp.add_run(); run.font.name='Garamond'; run.font.size=Pt(10)
for tag,val in [('w:fldChar','begin'),('w:instrText','PAGE'),('w:fldChar','end')]:
    el=OxmlElement(tag)
    if tag=='w:instrText': el.text=val
    else: el.set(qn('w:fldCharType'),val)
    run._r.append(el)

out=f"{OUTPUT_DIR}/{safe}_KDP.docx"
doc.save(out)
total_words=sum(len(p.text.split()) for p in doc.paragraphs)
print(f"\n🎉 BOOK COMPLETE!")
print(f"   File       : {out}")
print(f"   Total words: {total_words:,}")
print(f"   Est. pages : ~{total_words//250} pages (6×9 KDP)")
print(f"\n📂 Download from: Google Drive → AI_Book_Writer → output")
```
